```python
import numpy as np
import pandas as pd
 
np.random.seed(42)

cgpa = np.round(np.random.uniform(4, 10, 100), 2)
resume_score = np.round(np.random.uniform(1,11,100),2)

placed = ((cgpa*0.7 + resume_score*0.3)>6.5).astype(int)
df = pd.DataFrame({"cgpa":cgpa,"resume_score":resume_score,
                "placed":placed})

df.to_csv("placement.csv",index=False)

df.shape
```

```python
df.head()
```

```python
import seaborn as sns

sns.scatterplot(data=df,x="cgpa",y="resume_score",hue="placed")
```

```python
x = df.iloc[:,0:2]
y = df.iloc[:,-1]
print(type(x))
print(type(y))
```

```python
from sklearn.linear_model import Perceptron
p = Perceptron()

p.fit(x,y)
```

```python
p.coef_
```

```python
p.intercept_
```

```python
from mlxtend.plotting import plot_decision_regions

plot_decision_regions(x.values,y.values,clf =p, legend=2)
```
